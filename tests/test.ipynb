```python
import requests
from pprint import pprint
import re

url = "https://unstop.com/api/public/opportunity/search-result?opportunity=hackathons&page=1&per_page=18&oppstatus=open&usertype=students&domain=2&sortBy=&orderBy=&filter_condition=&undefined=true"

# params = {
#     "page": 1
# }

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
}

res = requests.get(url, headers=headers)
data = res.json()

def clean_html(raw_html):
    return re.sub('<.*?>', '', raw_html)


# 🔥 THIS IS THE MAIN PART YOU WERE MISSING
for item in data.get("data", {}).get("data", []):   # 👈 iterate over all hackathons

    output = {
        "id": item.get("id"),
        "url": item.get("seo_url"),
        "title": item.get("title"),
        "created_at": item.get("approved_date"),

        "short_description": clean_html(item.get("details", "")),

        "organization": item.get("organisation", {}).get("name"),

        "deadline": item.get("end_date"),
        "start_date": item.get("regnRequirements", {}).get("start_regn_dt"),

        "team_size": {
            "min": item.get("regnRequirements", {}).get("min_team_size"),
            "max": item.get("regnRequirements", {}).get("max_team_size")
        },

        "mode": item.get("region"),
        "is_paid": item.get("isPaid"),
        "tags": item.get("tags") or [],

        "skills": [
            s.get("skill_name")
            for s in item.get("required_skills", [])
            if s.get("skill_name")
        ],

        "location": item.get("address_with_country_logo", {}).get("city"),

        "participants": item.get("registerCount"),
        "views": item.get("viewsCount")
    }

    pprint(output)
    print("-" * 80)
```
